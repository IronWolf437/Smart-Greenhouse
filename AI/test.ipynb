```python
import matplotlib.pyplot as plt
from ultralytics import YOLO
import os
import cv2
import math


def show_single_result(result):
    for r in result:
        img = r.plot()
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        plt.figure(figsize=(8, 8))
        plt.imshow(img)
        plt.axis('off')


def show_multiple_results(results):
    total = len(results)
    cols = 5
    rows = math.ceil(total / cols)

    plt.figure(figsize=(16, 10))

    for idx, result in enumerate(results, 1):
        r = result[0]   # YOLO بيرجع list, فبنجيب أول عنصر
        img = r.plot()
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        plt.subplot(rows, cols, idx)
        plt.imshow(img)
        plt.axis('off')

    plt.tight_layout()
    plt.show()



model = YOLO('/media/ironwolf/study/هندسة/مشروع التخرج/AI/runs/detect/train/weights/best.pt')
path = input("Enter image or directory path: ")


if os.path.isfile(path):
    result = model(path)
    show_single_result(result)

elif os.path.isdir(path):
    results = []
    for i in os.listdir(path):
        img_path = os.path.join(path, i)
        result = model(img_path)
        results.append(result)

    show_multiple_results(results)

else:
    print("I am nobody")
```
